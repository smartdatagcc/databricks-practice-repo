# Configure Access to Azure SQL Database via Lakehouse Fedration
### 1.Create Connection: asql_connection_db
### 2.Create Foreign Catalog: asql_gizmobox_nara-db_catalog

```python
%sql
CREATE CONNECTION asql_gizmobox_nara_db_conn_sql
TYPE sqlserver
OPTIONS (
  host 'gizmobox-practice-server.database.windows.net',
  port '1433',
  user 'gizmoboxadmin',
  password 'Databricks@123'
);
```

```python
%sql
CREATE FOREIGN CATALOG IF NOT EXISTS asql_gizmobox_nara_Sql_Catalog
USING CONNECTION asql_gizmobox_nara_db_conn_sql
OPTIONS (
  database 'gizmobox-practice-db'
);

```

```python
%sql
SELECT * FROM asql_gizmobox_nara_sql_catalog.dbo.refundsnara
```

```python
%sql
SELECT *
FROM asql_gizmobox_nara_catalog_ui.dbo.refundsnara AS r
JOIN gizmobox_nara.bronze.payments AS p
ON r.payment_id = p.payment_Id;
```
