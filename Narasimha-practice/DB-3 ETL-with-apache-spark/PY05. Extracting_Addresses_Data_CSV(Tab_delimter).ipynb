# Extracting Addresses Data With TAB_DELIMETER
###   1. Read all the addresses files
###   2. Create addresses table in the bronze layer

### 1.Read all the addresses files

```python
df = spark.read.csv("/Volumes/gizmobox_nara/landing/operational_volume/addresses/")
display(df)
```

```python
df_format = (
      spark.read.format("csv")
      .option("header", "true")
      .option("delimiter","\t")
    .load("/Volumes/gizmobox_nara/landing/operational_volume/addresses/")
)
```

```python
display(df_format)
```

### Writing to Bronze layer

```python
df_format.writeTo("gizmobox_nara.bronze.py_addresses").createOrReplace()
```

```python
df_Py_addresses = spark.table("gizmobox_nara.bronze.py_addresses")
display(df_Py_addresses)
```
