```python
%sql
SELECT value:order_id AS order_id,
value
from gizmobox_nara.bronze.v_orders
```

### Extract Array Elements

```python
%sql
SELECT value:items[0] AS items,
value:items[1] AS items,
value
FROM gizmobox_nara.bronze.v_orders
```

### Extact Nested Column Values

```python
%sql
SELECT value:items[0].item_id As item_1_item_id,
       value:items[1].item_id As item_2_item_id,
       value
       FROM gizmobox_nara.bronze.v_orders
```

### CAST Column Values to a Specific Data Type

```python
%sql
SELECT value:items[0].item_id::INTEGER As item_1_item_id,
       value:items[1].item_id::INTEGER As item_2_item_id,
       value
       FROM gizmobox_nara.bronze.v_orders
```
