### Transform_MemberShip_Data
### 1.Extracting Customer_id from file Path
### 2. Transform to silver layer

```python
df_membership = spark.read.table("gizmobox_nara.bronze.v_memberships")
display(df_membership)
```

```python
%sql
SELECT * FROM gizmobox_nara.bronze.v_memberships
```

```python
import pyspark.sql.functions as f
df_membership_filepath = (
    df_membership
    .select(
        f.regexp_extract("path", '.*/([0-9]+).png$', 1).alias("customer_id"),
        "content" 
    )
)
display(df_membership_filepath)
```

```python
%sql
SELECT regexp_extract(path, '.*/([0-9]+).png$', 1) AS customer_id,content FROM gizmobox_nara.bronze.v_memberships
```

```python
df_membership_filepath.writeTo("gizmobox_nara.silver.py_memberships").createOrReplace()
```
