### RUN SQL COMMANDS USING PYTHON - Spark.SqlFunction

```python
df = spark.sql('SELECT * FROM json.`/Volumes/gizmobox_nara/landing/operational_volume/customers`')
display(df)
```

```python
spark.sql('''CREATE OR REPLACE Temp VIEW SPark_SQL
           AS 
          SELECT * FROM json.`/Volumes/gizmobox_nara/landing/operational_volume/customers`''')
```

```python
%sql
select * from SPark_SQL
```

### USING SPARK DATAFRAME READER API

```python
df= spark.read.format('json').load('/Volumes/gizmobox_nara/landing/operational_volume/customers')
display(df)
```

### USING SPARK TABLE FUNCTION 

```python
df = spark.table('gizmobox_nara.bronze.v_addresses')
display(df)
```
