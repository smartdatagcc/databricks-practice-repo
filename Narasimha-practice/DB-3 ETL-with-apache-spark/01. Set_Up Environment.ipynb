# CREATE _VOLUME_

```python
%sql
USE CATALOG gizmobox_nara;

-- Drop conflicting volume in default schema that occupies the same storage location
DROP VOLUME IF EXISTS gizmobox_nara.default.operational_data;

USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_volume
 LOCATION 'abfss://gizmobox-nara@ourdbpracticeextdl.dfs.core.windows.net/Landing/Operational_data/'
COMMENT 'operational data volume';

```
