# EXTRACT REFUND DATA FROM SQL TABLE(Currently not Hive_metastore) 
###  01. Read file from Azure SQL via JDBC
###  02. Creating External Table in Bronze Schema

## 01. Read file from Azure SQL via JDBC

```python
df = (
    spark.read.format('jdbc')
    .option('url', 'jdbc:sqlserver://gizmobox-practice-server.database.windows.net:1433;database=gizmobox-practice-db')
    .option('dbtable', 'RefundsNara')
    .option('user', 'gizmoboxadmin')
    .option('password', 'Databricks@123')
    .load()
)
display(df)
```

## 02. Write To Bronze Schema

```python
df.writeTo('gizmobox_nara.bronze.py_refunds').createOrReplace()
```

## Read the file from Bronze Schema

```python
df_read = spark.read.table('gizmobox_nara.bronze.py_refunds')
display(df_read)
```
