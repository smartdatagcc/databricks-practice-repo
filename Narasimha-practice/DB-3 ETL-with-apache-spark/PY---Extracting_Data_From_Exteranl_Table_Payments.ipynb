## Creating Schema for Header

```python
payment_schema = 'payment_Id integer,order_id Integer,payment_timestamp Timestamp,payment_status Integer,payment_method String'
```

### creating schema with pysaprk

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, TimestampType

py_payment_schema = StructType([
    StructField('payment_Id', IntegerType(), True),
    StructField('order_id', IntegerType(), True),
    StructField('payment_timestamp', TimestampType(), True),
    StructField('payment_status', IntegerType(), True)])
```

```python
df = (spark.read.format('csv')
     .option("delimiter",",")
     .schema(py_payment_schema)
     .load('abfss://gizmobox-nara@ourdbpracticeextdl.dfs.core.windows.net/Landing/External_data/Payments')
)
display(df)
```

# CREATING EXTERNAL TABLE ON DATA


```python
df.writeTo("gizmobox_nara.bronze.py_payments").createOrReplace()
```

```python
%sql
CREATE TABLE IF NOT EXISTS gizmobox_nara.bronze.payments
(payment_Id integer,order_id Integer,payment_timestamp Timestamp,payment_status Integer,payment_method String)
USING CSV
OPTIONS (
    header ='true',
    delimiter = ","
)
LOCATION 'abfss://gizmobox-nara@ourdbpracticeextdl.dfs.core.windows.net/Landing/External_data/Payments'
```
