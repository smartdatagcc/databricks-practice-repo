# Extracting the Order File from Complex_JSON
### 1.Query the file from JSON Format
### 2.Query the file from Text Format
### 3.Create Order Views in Bronze Schema

```python
df = spark.read.format('json').load("/Volumes/gizmobox_nara/landing/operational_volume/orders")
display(df)
```

```python
display(df)
```

```python
from pyspark.sql import functions as F
df_with_metadata = df.select(F.col("_metadata.file_path"), "*")
```

```python
%sql
SELECT *, _metadata.file_path As filepath FROM json.`/Volumes/gizmobox_nara/landing/operational_volume/orders/`
```

# Query The Order Records Using Text Format


```python
df = spark.read.text("/Volumes/gizmobox_nara/landing/operational_volume/orders/")

display(df)
```

```python
%sql
SELECT * FROM text.`/Volumes/gizmobox_nara/landing/operational_volume/orders/`
```

## WRITE TO Bronze Schema

```python
df.writeTo("gizmobox_nara.bronze.py_orders").createOrReplace()
```

```python
df = spark.table("gizmobox_nara.bronze.py_orders")
display(df)
```

```python
%sql
SELECT * FROM gizmobox_nara.bronze.py_orders
```
