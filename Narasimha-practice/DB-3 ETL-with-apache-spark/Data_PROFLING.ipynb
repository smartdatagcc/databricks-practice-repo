Data Profling ![Manually](path)

```python
%sql
SELECT COUNT(*),COUNT(customer_id),COUNT(telephone),COUNT(email) FROM gizmobox_nara.bronze.v_customers
```

# COUNT_IF

```python
%sql
SELECT COUNT_IF(customer_id IS NULL),COUNT_IF(telephone IS NULL),COUNT_IF(email IS NULL) FROM gizmobox_nara.bronze.v_customers
```

# WHERE CLAUSE AND DISTINCT KEYWORD

```python
%sql
SELECT COUNT(customer_id) FROM gizmobox_nara.bronze.v_customers
WHERE customer_id IS NULL
```

```python
%sql
SELECT COUNT(*) AS total,
COUNT(DISTINCT customer_id) AS UNIQUECUSTOMERS FROM gizmobox_nara.bronze.v_customers
```

```python
%sql
SELECT COUNT(*) AS total,
COUNT(DISTINCT customer_id) AS UNIQUECUSTOMERS FROM gizmobox_nara.bronze.v_customers
WHERE customer_id IS NULL
```
