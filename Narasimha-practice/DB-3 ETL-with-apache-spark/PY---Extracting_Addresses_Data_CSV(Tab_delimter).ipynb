### 1.LIMIATION Of SELECT Statement in CSV_FILES

```python
df = spark.read.csv("/Volumes/gizmobox_nara/landing/operational_volume/addresses/")
```

```python
df = (spark.read.format("csv")
.option("header", "true")
.option("delimiter","\t")
.load("/Volumes/gizmobox_nara/landing/operational_volume/addresses/")
)
```

```python
display(df)
```

```python
%sql
SELECT * FROM csv.`/Volumes/gizmobox_nara/landing/operational_volume/addresses/`
```

# 2.USING TABLED VALUED FUNCTION

```python
%sql
SELECT * FROM read_files('/Volumes/gizmobox_nara/landing/operational_volume/addresses/',
                          format=>'csv',
                          delimiter=>'\t',
                          header=> True);
```

```python
df.writeTo("gizmobox_nara.bronze.py_addresses").createOrReplace()
```

```python
%sql
select * from gizmobox_nara.bronze.py_addresses
```

# CREATE A VIEW FOR ADRESSES IN BRONZE LAYER

```python
%sql
CREATE OR REPLACE VIEW gizmobox_nara.bronze.v_addresses AS
SELECT * FROM read_files('/Volumes/gizmobox_nara/landing/operational_volume/addresses/',
                          format=>'csv',
                          delimiter=>'\t',
                          header => true);
```

```python
%sql
SELECT * FROM gizmobox_nara.bronze.v_addresses
```
