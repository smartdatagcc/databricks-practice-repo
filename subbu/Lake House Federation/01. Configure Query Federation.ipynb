# Configure Access to Azure SQL Database via Lakehouse Fedration
### 1.Create Connection: asql_connection_db
### 2.Create Foreign Catalog: asql_gizmobox_subbu-db_catalog

```sql
%sql
CREATE CONNECTION asql_gizmobox_subbu_db_conn_sql
TYPE sqlserver
OPTIONS (
  host 'gizmobox-practice-server.database.windows.net',
  port '1433',
  user 'gizmoboxadmin',
  password 'Databricks@123'
);
```

```sql
%sql
CREATE FOREIGN CATALOG IF NOT EXISTS asql_gizmobox_subbu_Sql_Catalog
USING CONNECTION asql_gizmobox_subbu_db_conn_sql
OPTIONS (
  database 'gizmobox-practice-db'
);

```

```sql
SELECT *
FROM asql_gizmobox_subbu_Sql_Catalog.dbo.refunds
```

```sql
select * from gizmobox_catalog_subbu.bronze.payments
```

```sql
SELECT *
FROM asql_gizmobox_subbu_Sql_Catalog.dbo.refunds AS r
JOIN gizmobox_catalog_subbu.bronze.payments AS p
ON r.payment_id = p.payment_Id;
```
