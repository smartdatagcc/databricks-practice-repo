#### Extract Data From the Adress Files
1. Demonstrate Limitations of CSV file format in SELECT statement
2. Use read_files funtion to overcome the limitations
3. Create Addresses View in Bronze Schema

##### 1. Demonstrate Limitations of CSV file format in SELECT statement

```python
%sql
Select * from csv.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/addresses`
```

##### 2. Use read_files funtion to overcome the limitations

##### [Function Documentation](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/read_files)

```python
%sql
select * from read_files('/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/addresses',
                            format => 'csv',
                            header => True,
                            delimiter => '\t'
                           );
```

##### 3. Create Addresses View in Bronze Schema

```python
%sql
CREATE or REPLACE view gizmobox_catalog_subbu.bronze.v_address as select * from read_files('/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/addresses',
                            format => 'csv',
                            header => True,
                            delimiter => '\t'
                           );
```

```python
%sql
select * from gizmobox_catalog_subbu.bronze.v_address
```
