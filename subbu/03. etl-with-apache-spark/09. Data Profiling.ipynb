1. Profile Data using UI
2. Profile Data using DBUTILS Package(dbutils.data.summarize method)
3. Profile data manually
 
        - Count
        - Count_If
        - Min
        - Max
        - Where Clause

##### 1. Profile Data using UI


```sql
select * from gizmobox_catalog_subbu.bronze.v_customers
```

##### 2. Profile Data using DBUTILS Package(dbutils.data.summarize method)

```sql
%python
df=spark.table('gizmobox_catalog_subbu.bronze.v_customers')
dbutils.data.summarize(df) 
```

##### 3. Profile data manually

##### 3.1 COUNT Function

```sql
SELECT count(*),count(1),count(customer_id),count(customer_name),count(date_of_birth),count(email),count(member_since),count(telephone) from gizmobox_catalog_subbu.bronze.v_customers
```

##### 3.2 COUNT_IF Function

```sql
SELECT count(*),count(1),count_if(customer_id IS NULL),count_if(customer_name IS NULL),count_if(date_of_birth IS NULL),count_if(email IS NULL),count_if(member_since IS NULL),count_if(telephone IS NULL),count(customer_name),count(date_of_birth),count(email),count(member_since),count(telephone) from gizmobox_catalog_subbu.bronze.v_customers
```

##### 3.3 WHERE Clause

```sql
SELECT count(*) from gizmobox_catalog_subbu.bronze.v_customers
where customer_id is null
```

##### 3.4 DISTINCT Keyword

```sql
SELECT count(*) as Total_number_of_records,
count(distinct customer_id) as Total_number_of_distinct_records
from gizmobox_catalog_subbu.bronze.v_customers
 where customer_id is not null
```
