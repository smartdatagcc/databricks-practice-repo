#### Extract Data From the Adress Files
1. Read all the address files
3. Create Addresses table in Bronze Schema

##### 1. Read all the address files

```python
df=(
    spark.read.format('csv')
         .option('header',True)
         .option('delimiter','\t')
         .load('/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/addresses')    
)
display(df)
```

##### 3. Create Addresses table in Bronze Schema

```python
df.writeTo('gizmobox_catalog_subbu.bronze.py_addresses').createOrReplace()
```
