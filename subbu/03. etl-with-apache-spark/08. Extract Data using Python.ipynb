#### Extract Data - Using PySpark
1. Run SQL Commands using Python - spark.sql Funtion
2. Spark Dataframe Reader API
3. Read Tables using spark.table Function

##### 1. Run SQL Commands using Python - spark.sql Funtion


```python
df = spark.sql('SELECT * from JSON.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/`')
display(df)
```

```python
df = spark.sql('''CREATE OR REPLACE TEMP VIEW tv_customers 
               AS 
               SELECT * from JSON.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/`''')
```

```python
%sql
select * from tv_customers;
```

##### 2. Spark Dataframe Reader API
[DataframeReader API](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrameReader.html)

```python
df=spark.read.format("json").load("/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/")
display(df)
```
