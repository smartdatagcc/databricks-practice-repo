#### User Defined Functions (UDF)
- User Defined Functions (UDF) in Spark are custom functions created by user to extend the capabilities od Spark SQL and PySpark
- UDFs allow us to perform calculations and transforamtions to apply business logic that are not possible with built-in functions
- You define the functions once and use it across multiple queries
- Sql UDFs are recommended over Python UDFs due to better optimization

#### Syntax
---------------------------------------

**CREATE FUNCTION** Catalog_name.Schema_name.Udf_name(param_name data_type)

**RETURNS** return_type

**RETURN** expression;

```sql
CREATE or REPLACE FUNCTION gizmobox_catalog_subbu.default.get_fullname(firstname String, surname String)
RETURNS STRING
return concat(initcap(firstname), ' ',initcap(surname));
```

```sql
SELECT gizmobox_catalog_subbu.default.get_fullname('Subba','Reddy')
```

```sql
DESC FUNCTION EXTENDED gizmobox_catalog_subbu.default.get_fullname
```

```sql
CREATE or REPLACE FUNCTION gizmobox_catalog_subbu.default.get_payment_status(payment_status INTEGER)
RETURNS STRING
RETURN CASE payment_status
    WHEN 1 THEN 'Success'
    WHEN 2 THEN 'Pending'
    WHEN 3 THEN 'Cancelled'    
    WHEN 4 THEN 'Failed'    
  END;
```

```sql
SELECT
  payment_id, order_id,
  payment_timestamp,
  CAST(date_format(payment_timestamp, 'yyyy-MM-dd') as date) as payment_date,
  date_format(payment_timestamp, 'HH:mm:ss') as payment_time,
  gizmobox_catalog_subbu.default.get_payment_status(payment_status) as payment_status,
  payment_method
FROM
  gizmobox_catalog_subbu.bronze.payments
```
