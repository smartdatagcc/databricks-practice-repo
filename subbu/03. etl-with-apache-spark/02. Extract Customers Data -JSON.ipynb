1. Querying single file
2. Query List of Files using wildcard Charecters
3. Query all the files in a folder

##### 1. Querying single file

```sql
%fs ls /Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/
```

```sql
select * from json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/customers_2024_10.json`
```

##### 2. Query List of Files using wildcard Charecters

```sql
select * from json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/customers_2024_*.json`
```

##### 3. Query all the files in a folder

```sql
select * from json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/`
```

##### 4. Select file MetaData

```sql
select input_file_name() as file_path, _metadata, _metadata.file_path  as file_path,* from json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/`
```

##### 5. Register files in Unity Catalog using Views

```sql
Create or replace view gizmobox_catalog_subbu.bronze.v_customers as
select _metadata.file_path  as file_path,* from json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/`
```

```sql
select * from gizmobox_catalog_subbu.bronze.v_customers
```

##### 6. Create Temporary View

```sql
Create or replace temporary view tv_customers as
select _metadata.file_path  as file_path,* from json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/`
```

```sql
select * from tv_customers
```

#####7. Create Global Temporary View


```sql
Create or replace global temporary view gtv_customers as
select _metadata.file_path  as file_path,* from json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/`
```

```sql
select * from global_temp.gtv_customers;
```
