#### Extract Data From Orders Data - Complex JSON
1. Querying Orders File using JSON Format
2. Query Orders File using TEXT Format
3. Create Orders Table in Bronze Schema

##### 1. Querying Orders File using JSON Format

```python
df = spark.read.json("/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/orders/");
display(df)
```

##### 2. Query Orders File using TEXT Format

```python
df = spark.read.text("/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/orders/")
display(df)
```

##### 3. Create Orders View in Bronze Schema

```python
df.writeTo('gizmobox_catalog_subbu.bronze.py_orders').createOrReplace()
```
