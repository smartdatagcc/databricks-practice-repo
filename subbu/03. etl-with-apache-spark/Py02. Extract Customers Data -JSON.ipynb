##### 3. Query all the files in a folder

```python
df = spark.read.format("json").load("/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/")
display(df)
```

```python
df = spark.read.json("/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers/")
display(df)
```

##### 4. Select file MetaData

```python
df_meta_data = df.select("_metadata.file_path","*")
display(df_meta_data)
```

##### 5. Create Table in Bronze Schema

```python
df_meta_data.write.format("delta").mode("overwrite").saveAsTable("gizmobox_catalog_subbu.bronze.py_customers")
```

```python
df_meta_data.writeTo("gizmobox_catalog_subbu.bronze.py_customers").createOrReplace()
```
