#### Monthly Order Summary
for each of the custmoer, produce the following summary per month

1. total orders
2. total items bought
3. total amount spent

```sql
select * from gizmobox_catalog_subbu.silver.orders
```

```sql
Select date_format(transaction_timestamp,'yyyy-MM') as order_month,
customer_id,
count(distinct order_id) as total_orders,
sum(quantity) as total_items_bought,
sum(price*quantity) as total_amount
from 
gizmobox_catalog_subbu.silver.orders
Group by order_month,customer_id
```

```sql
CREATE Table if not exists gizmobox_catalog_subbu.gold.order_summary_monthly
As
Select date_format(transaction_timestamp,'yyyy-MM') as order_month,
customer_id,
count(distinct order_id) as total_orders,
sum(quantity) as total_items_bought,
sum(price*quantity) as total_amount
from 
gizmobox_catalog_subbu.silver.orders
Group by order_month,customer_id
```
