#### Transform Addresses Data

1. Create one record for each customer with 2 sets of address columns, 1 for shipping and 1 for billing address
2. Write transformed data to the Silver schema

```sql
select customer_id,
    address_type,
    address_line_1,
    city,
    state,
    postcode
from gizmobox_catalog_subbu.bronze.v_address;
--describe gizmobox_catalog_subbu.bronze.v_address;
```

##### 1. Create one record for each customer with 2 sets of address columns, 1 for shipping and 1 for billing address


```sql

SELECT *
FROM ( select customer_id,
        address_type,
        address_line_1,
        city,
        state,
        postcode
    from gizmobox_catalog_subbu.bronze.v_address)
PIVOT(
    MAX(address_line_1) as address_line_1,
    MAX(city) as city,
    MAX(state) as state,
    MAX(postcode) as postcode
    FOR address_type IN ('shipping', 'billing')
);
```

##### 2. Write transformed data to the Silver schema

```sql
CREATE TABLE gizmobox_catalog_subbu.silver.addresses
SELECT *
FROM ( select customer_id,
        address_type,
        address_line_1,
        city,
        state,
        postcode
    from gizmobox_catalog_subbu.bronze.v_address)
PIVOT(
    MAX(address_line_1) as address_line_1,
    MAX(city) as city,
    MAX(state) as state,
    MAX(postcode) as postcode
    FOR address_type IN ('shipping', 'billing')
);
```
