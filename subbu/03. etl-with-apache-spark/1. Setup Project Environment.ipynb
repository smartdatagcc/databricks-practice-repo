## Setup the project environment for GizomoBox Data Lakehouse

#### 1. Access the container for gizmobox



```sql
%fs ls 'abfss://gizmoboxsubbu@ourdbpracticeextdl.dfs.core.windows.net/landing'
```

#### 2. Create external location

```sql
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_practice_ext_dl_gizmoboxsubbu
    URL 'abfss://gizmoboxsubbu@ourdbpracticeextdl.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-practice-ext-sc`)
    COMMENT 'External Location for GizmoBox Data Lakehouse'
```

#### 3. Create a catalog - Gizmobox

```sql
%sql
CREATE CATALOG IF NOT EXISTS gizmobox_catalog_subbu
    MANAGED LOCATION 'abfss://gizmoboxsubbu@ourdbpracticeextdl.dfs.core.windows.net/gizmoboxsubbu'
    COMMENT 'This is the catalog for the gizmobox-subbu data lake house';
```

##### See the existing catalogs

```sql
%sql
SHOW CATALOGS
```

#### 4. Create Schemas
1. Landing
4. Bronze
1. Silver
1. Gold

```sql
%sql
Select current_catalog()
```

```sql
%sql
USE CATALOG gizmobox_catalog_subbu;
SHOW SCHEMAS
```

```sql
%sql
USE CATALOG gizmobox_catalog_subbu;
CREATE SCHEMA IF NOT EXISTS landing
    MANAGED LOCATION  'abfss://gizmoboxsubbu@ourdbpracticeextdl.dfs.core.windows.net/gizmoboxsubbu/landing'
    COMMENT 'This is the landing schema for the gizmobox-subbu data lake house';
  
CREATE SCHEMA IF NOT EXISTS bronze
    MANAGED LOCATION  'abfss://gizmoboxsubbu@ourdbpracticeextdl.dfs.core.windows.net/gizmoboxsubbu/bronze'
    COMMENT 'This is the bronze schema for the gizmobox-subbu data lake house';

CREATE SCHEMA IF NOT EXISTS silver
    MANAGED LOCATION  'abfss://gizmoboxsubbu@ourdbpracticeextdl.dfs.core.windows.net/gizmoboxsubbu/silver'
    COMMENT 'This is the silver schema for the gizmobox-subbu data lake house';

CREATE SCHEMA IF NOT EXISTS gold
    MANAGED LOCATION  'abfss://gizmoboxsubbu@ourdbpracticeextdl.dfs.core.windows.net/gizmoboxsubbu/gold'
    COMMENT 'This is the gold schema for the gizmobox-subbu data lake house';
  
```

#### 5. Create Volume

```sql
%sql
USE CATALOG gizmobox_catalog_subbu;
USE SCHEMA  landing;


CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data_volume
    COMMENT 'This is the operational data volume for the gizmobox-subbu data lake house'
    LOCATION 'abfss://gizmoboxsubbu@ourdbpracticeextdl.dfs.core.windows.net/landing/operational_data/';
```
