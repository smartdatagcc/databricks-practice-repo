#### Query Orders Data as JSON Strings
1. Extract Top level Column Values
2. Extract Array Elements
3. Extract Nested Column Values
4. CAST column values to Specific Data Type

```sql
SELECT * FROM gizmobox_catalog_subbu.bronze.v_orders
```

##### 1. Extract Top level Column Values

```sql
SELECT value:order_id as order_id, value:customer_id as customer_id,
    value
 FROM gizmobox_catalog_subbu.bronze.v_orders
```

##### 2. Extract Array Elements

```sql
SELECT value:items[0] as items_1, 
       value:items[1] as items_2,
       value
 FROM gizmobox_catalog_subbu.bronze.v_orders
```

##### 3. Extract Nested Column Values

```sql
SELECT value:items[0].item_id as items_1_item_id, 
       value:items[0] as items_1,
       value:items[1] as items_2,
       value
 FROM gizmobox_catalog_subbu.bronze.v_orders
```

##### 4. CAST column values to Specific Data Type

```sql
SELECT 
      value:items[0].item_id::INTEGER as items_1_item_id, 
      --cast(value:items[0].item_id as INTEGER) as items_1_item_id,
       value:items[0] as items_1,
       value:items[1] as items_2,
       value
 FROM gizmobox_catalog_subbu.bronze.v_orders
```
