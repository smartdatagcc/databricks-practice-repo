### Insert Overwrite

```sql
%sql
drop table if exists gizmobox_sivan.delta_lake.gold_companies;
Create table gizmobox_sivan.delta_lake.gold_companies(
    company_name string,
    founded_date date,
    country string
);

insert into gizmobox_sivan.delta_lake.gold_companies
(company_name,founded_date, country)
values
('Google', '1996-01-01', 'USA'),
('Amazon', '1998-01-01', 'China');
```

```sql
%sql
select * from gizmobox_sivan.delta_lake.gold_companies
```

```sql
%sql
drop table if exists gizmobox_sivan.delta_lake.bronze_companies;
Create table gizmobox_sivan.delta_lake.bronze_companies(
    company_name string,
    founded_date date,
    country string
);

insert into gizmobox_sivan.delta_lake.bronze_companies
(company_name,founded_date, country)
values
('Microsoft', '2019-02-01', 'USA'),
('Google', '1996-01-01', 'USA'),
('Apple', '1986-01-01', 'USA'),
('Tencent', '1994-01-01', 'USA'),
('Amazon', '1998-01-01', 'China');
```

```sql
%sql
select * from gizmobox_sivan.delta_lake.bronze_companies;
```

```sql
%sql
insert overwrite table gizmobox_sivan.delta_lake.gold_companies
select * from gizmobox_sivan.delta_lake.bronze_companies;
```

```sql
%sql
select * from gizmobox_sivan.delta_lake.gold_companies;
```

```sql
desc history gizmobox_sivan.delta_lake.gold_companies;
```

**2. Replace all the data from specific partition**

```sql
Create table gizmobox_sivan.delta_lake.gold_companies_partitioned(
    company_name string,
    founded_date date,
    country string
)
partitioned by (country);

insert into gizmobox_sivan.delta_lake.gold_companies_partitioned
(company_name,founded_date, country)
values
('Google', '1996-01-01', 'USA'),
('Amazon', '1998-01-01', 'China');
```

```sql
select * from gizmobox_sivan.delta_lake.gold_companies_partitioned;
```

```sql
desc extended gizmobox_sivan.delta_lake.gold_companies_partitioned;
```

```sql
drop table if exists gizmobox_sivan.delta_lake.bronze_companies_USA;
Create table gizmobox_sivan.delta_lake.bronze_companies_USA(
    company_name string,
    founded_date date,
    country string
);

insert into gizmobox_sivan.delta_lake.bronze_companies_USA
(company_name,founded_date, country)
values
('Microsoft', '2019-02-01', 'USA'),
('Google', '1996-01-01', 'USA'),
('Apple', '1986-01-01', 'USA'),
('Tencent', '1994-01-01', 'USA');
```

```sql
select * from gizmobox_sivan.delta_lake.bronze_companies_USA;
```

```sql
insert overwrite table gizmobox_sivan.delta_lake.gold_companies_partitioned
partition(country='USA')
select 
    company_name,
    founded_date
from gizmobox_sivan.delta_lake.bronze_companies_USA;
```

```sql
select * from gizmobox_sivan.delta_lake.gold_companies_partitioned;
```

```sql
desc history gizmobox_sivan.delta_lake.gold_companies_partitioned;
```

**3.How to handle schema changes**

insert overwrite --> Use to override the data in a table or partition wheren there are no schema changes\
Create or replace table --> Whenever the schema changes
