### Understanding Delta Lake Transaction Log

Understanding the cloud storage directory structure behind delta lake tables

**0.Create a new schema under the demo catalog for this section of the course (delta_lake)**

```python
%sql
CREATE SCHEMA IF NOT EXISTS gizmobox_sivan.delta_lake
MANAGED LOCATION 'abfss://gizmoboxsivan@ourdbpracticeextdl.dfs.core.windows.net/delta_lake'
```

**1. Create a Delta Lake table**

```python
%sql
CREATE TABLE IF NOT EXISTS gizmobox_sivan.delta_lake.companies
(
    company_name string, 
    founded_date date,
    country string
)
```
