### Remove unused files using VACUUM

####VACUUM Command
Used to remove old, unused files from the Delta Lake to free up the storage. Permanently delete the files that are no longer referenced in the transaction log 

#### 1. Check the table history

```sql
DESC HISTORY demo.delta_lake.optimize_stock_prices;
```

#### 2. Run VACUUM

```sql
vacuum demo.delta_lake.optimize_stock_prices;
```

```sql
DESC HISTORY demo.delta_lake.optimize_stock_prices;
```

```sql
-- SET spark.daatabricks.delta.retentionDurationCheck.enabled = FALSE
VACUUM demo.delta_lake.optimize_stock_prices RETAIN 20 HOURS;
```

```sql
DESC HISTORY demo.delta_lake.optimize_stock_prices;
```
