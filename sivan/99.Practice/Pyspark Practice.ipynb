```python
from pyspark.sql.functions import when

empl_df.withColumn(
    "salary_category",
    when (col("salary") >= 60000, "high")
    .otherwise ("normal")
).show()
```

```python
empl_df = empl_df.withColumn(
    "slaray_category",
    when(col("salary") >= 60000, "high")
    .when(col("salary") > 45000, "medium")
    .otherwise("low")
)

display(empl_df)
```

```python
employees = [
    (1, "John", "IT", 50000, "Hyderabad"),
    (2, "David", "HR", 40000, "Bangalore"),
    (3, "Smith", "IT", 60000, "Hyderabad"),
    (4, "Peter", "Finance", 55000, "Chennai"),
    (5, "Robert", "HR", 45000, "Bangalore"),
    (6, "James", "IT", 70000, "Pune"),
    (7, "David", "HR", 40000, "Bangalore")
]

columns = [
    "emp_id", "emp_name", "department",
    "salary", "city"
]

emp2_df = spark.createDataFrame(employees, columns)
```

**1. Display only emp_name, department and salary.**

```python
emp2_df.select(
    col("emp_name"),
    col("department"),
    col("salary")
).show()
```

**2.Retrieve employees whose salary is greater than 50000.**

```python
emp2_df.where(
    col("salary") > 50000
).show()
```

**3.Retrieve employees who belong to IT and whose salary is greater than 50000.**

```python
emp2_df.filter(
    (col("department") == "IT") &
    (col("salary") > 50000)
).show()
```

**4.Create a new column called bonus, calculated as 10% of salary.**

```python
emp2_df.withColumn(
    "bonus",
    col("salary") * 0.1
).show()
```

**5.Create a new column called employee_type. If salary is greater than or equal to 55000, assign Senior; otherwise assign Junior.**

```python
emp2_df.withColumn(
    "employee_type",
    when(col("salary") >= 55000, "senior")
    .otherwise("junior")
).show()
```

**6.Remove the city column.**

```python
emp2_df.drop(col("city")).show()
```

**7.Remove completely duplicated rows.**

```python
emp2_df.distinct().show()
```

**8.Remove duplicates based on emp_name.**

```python
emp2_df.dropDuplicates(
    ["emp_name"]
).show()
```

**9.Display only the first three records.**

```python
emp2_df.limit(3).show()
```

**10. Sort employees by salary in descending order.**

```python
emp2_df.orderBy(
    col("salary").desc()
).show()
```
