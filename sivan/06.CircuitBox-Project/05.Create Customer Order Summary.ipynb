%md
##### Gold_Customer_Order_Summary (Order summary of each customer along with their latest address details)
 
 
        --> Silver_customers --> Customer_id
                             --> Customer_name
                             --> date_of_birth
                             --> telephone
                             --> email
                       
        --> silver_addresses --> address_line_1        ( Get the address at the time of order being placed)
                             --> city
                             --> state
                             --> postcode
 
        --> silver_orders    --> total_orders          (Aggregate Values)
                             --> total_items_ordered
                             --> total_order_amount
 

```sql
CREATE OR REFRESH MATERIALIZED VIEW gold_customer_order_summary
AS
select
    c.customer_id,
    c.customer_name,
    c.date_of_birth,
    c.email,
    c.telephone,
    a.address_line_1,
    a.city,
    a.state,
    a.postcode,
    count(distinct o.order_id) as total_orders,
    sum(o.item_quantity) as total_quantity,
    sum(o.item_price * o.item_quantity) as total_order_amount
from  live.silver_customers as c
    join live.silver_addresses as a
        on c.customer_id = a.customer_id
    join live.silver_orders as o
        on c.customer_id = o.customer_id
where a.__END_AT is null
group by All;    
```
