### 2. Use read_Files function to overcome the limitations

```python
df= (
    spark.read.format("csv")
        .option("header","true")
        .option("delimiter","\t")
        .load("/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/addresses")
)
display(df)
```

[Function Documentation](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/read_files)

### 3. Create Addresses table in Bronze layer

```python
df.writeTo("gizmobox_sivan.bronze.py_addresses").createOrReplace();
```
