### Monthly Order Summary

For each of the customer, produce the following summary per month
 1. total oders
 2. total items bought
 3. total amount spent

```python
dfOrders = spark.table("gizmobox_sivan.silver.orders")
dfOrders.display()
```

```python
from pyspark.sql import functions as f

dfOrder_monthly_Summary = (
    spark.table("gizmobox_sivan.silver.orders")
    .groupBy(
        f.date_format(f.col("transaction_timestamp"), "yyyy-MM").alias("order_month"),
        f.col("customer_id")
    )
    .agg(
        f.countDistinct("order_id").alias("total_orders"),
        f.sum("quantity").alias("total_items"),
        f.sum(f.col("price") * f.col("quantity")).alias("total_amount")
    )
)

dfOrder_monthly_Summary.display()
```

```python
dfOrder_monthly_Summary.writeTo("gizmobox_sivan.gold.py_order_summary_monthly").createOrReplace()
```
