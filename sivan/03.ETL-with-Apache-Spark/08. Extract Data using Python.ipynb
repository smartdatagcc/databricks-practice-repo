### 1. Run SQL Commands using Python - spark.sql function

```python
df = spark.sql('select * from json.`/Volumes/gizmobox_sivan/landing/operational_data_volume/customers`')
display(df)
```

```python
spark.sql('''
CREATE OR REPLACE TEMP VIEW tv_customers AS
SELECT * FROM json.`/Volumes/gizmobox_sivan/landing/operational_data_volume/customers`''')
```

```python
%sql
select * from tv_customers;
```

### 2. Spark Dataframe Reader API
[Documentation](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrameReader.html)

```python
df = spark.read.format("json").load("/Volumes/gizmobox_sivan/landing/operational_data_volume/customers")
display(df)

```

### 3. spark.table Function

```python
df = spark.table("gizmobox_sivan.bronze.v_customers")
display(df)
```
