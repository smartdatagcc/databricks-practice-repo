### Extract Data from the Address Files
 1. Demonstrate Limitations of CSV file format in SELECT statement
 2. Use read_Files function to overcome the limitations
 3. Create Adderesses view in Bronze Layer

### 1. Demonstrate Limitations of CSV file format in SELECT statement

```sql
%fs ls '/Volumes/gizmobox_sivan/landing/operational_data_volume/addresses/'

```

```sql
select * from csv.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/addresses`
```

### 2. Use read_Files function to overcome the limitations

[Function Documentation](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/read_files)

```sql
select *
from read_files('/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/addresses',
                format => 'csv',
                header => True,
                delimiter=> '\t')
```

### 3. Create Addresses view in Bronze layer

```sql
CREATE OR REPLACE VIEW gizmobox_sivan.bronze.v_addresses 
AS
select *
from read_files('/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/addresses',
                format => 'csv',
                delimiter => '\t',
                header => True)
```

```sql
select * from gizmobox_sivan.bronze.v_addresses
```
