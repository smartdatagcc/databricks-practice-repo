### Configure Access to Azure SQL Database via Lakehouse Federation
- Create connection - asql_gizmobox_sivan_db_conn
- Create Foreign Catalog - asql_gizmobox_sivan_db_catalog

```sql
CREATE CONNECTION asql_gizmobox_sivan_db_conn
TYPE sqlserver
OPTIONS (
  host 'gizmobox-practice-server.database.windows.net',
  port '1433',
  user 'gizmoboxadmin',
  password 'Databricks@123'
);
```

```sql
CREATE FOREIGN CATALOG IF NOT EXISTS asql_gizmobox_sivan_db_catalog
USING CONNECTION asql_gizmobox_subbu_db_conn_sql
OPTIONS (
  database 'gizmobox-practice-db'
);
```
