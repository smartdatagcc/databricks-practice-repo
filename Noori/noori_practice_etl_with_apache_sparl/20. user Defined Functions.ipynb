##### User Defined Functions (UDF)
###### - user defined functions (UDF) in spark are custom fuctions created by users to extend the capabilities of Spark SQL and PySpark
###### - UDFs allow us to perform calculations or tranformations to apply business logic that are not possible with built-in functions
###### - you define the functions once and use it accross multiple queries
###### - SQL UDFs are recommended over python UDFs due to better optimization

### Write a funtion to concatenate firstname and surname to and output to fullname

```sql
create or replace function gizmobox_catalog_noori.default.get_customer_fullname(firstname string, lastname string) 
returns string 
return concat(initcap(firstname), ' ',initcap(lastname));
```

```sql
select gizmobox_catalog_noori.default.get_customer_fullname('john','doe')
```
