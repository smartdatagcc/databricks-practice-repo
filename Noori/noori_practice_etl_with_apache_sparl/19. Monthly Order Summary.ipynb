### Monthly order summary
#### For each of the customer, produce the following summary per month
##### 1. total orders
##### 2. total items bought
##### 3. total amount spent

```sql
select 
 *

from gizmobox_catalog_noori.silver.orders
```

```sql
create table if not exists gizmobox_catalog_noori.gold.order_summary_montly
as

select 
 date_format(transaction_timestamp, 'yyyy-MM') as order_month,
 customer_id,
 count(Distinct order_id) as total_orders,
 sum(quantity) as total_items_count,
 sum(price * quantity) as total_revenue

from gizmobox_catalog_noori.silver.orders
group by customer_id, order_month
```

```sql
select * from gizmobox_catalog_noori.gold.order_summary_montly
```
