# 05. Extract Addresses Data using Read_files Function - tsv

## Limitation of csv file format in select statement

```sql
select * from csv.`/Volumes/gizmobox_catalog_noori/landing/vol_operational_data/addresses/`
```

## 2. Use read_files functon to overcome the limitation

```sql
select * from 
read_files('/Volumes/gizmobox_catalog_noori/landing/vol_operational_data/addresses/',
format => 'csv',
delimiter => '\t',
header => 'true'
)
```

```sql
create or replace view gizmobox_catalog_noori.bronze.v_addresses as
select * from 
read_files('/Volumes/gizmobox_catalog_noori/landing/vol_operational_data/addresses/',
format => 'csv',
delimiter => '\t',
header => 'true'
)
```

```sql
select * from v_addresses
```
