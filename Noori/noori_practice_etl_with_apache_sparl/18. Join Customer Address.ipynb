### Join Customer and Address
##### join customer data with address data to create a customer_address table which contains the address of each customer on the same record

```sql
select 
  c.customer_id, 
  c.customer_name, 
  c.date_of_birth, 
  c.email, 
  c.member_since, 
  c.telephone, 
  a.shipping_address_line_1, 
  a.shipping_city, 
  a.shipping_state, 
  a.shipping_postalcode, 
  a.billing_address_line_1, 
  a.billing_city, 
  a.billing_state, 
  a.billing_postalcode 
from gizmobox_catalog_noori.silver.customers c
join gizmobox_catalog_noori.silver.addresses a on a.customer_id = c.customer_id
```

```sql
CREATE TABLE IF NOT EXISTS gizmobox_catalog_noori.gold.customers_addresses
AS
select 
  c.customer_id, 
  c.customer_name, 
  c.date_of_birth, 
  c.email, 
  c.member_since, 
  c.telephone, 
  a.shipping_address_line_1, 
  a.shipping_city, 
  a.shipping_state, 
  a.shipping_postalcode, 
  a.billing_address_line_1, 
  a.billing_city, 
  a.billing_state, 
  a.billing_postalcode 
from gizmobox_catalog_noori.silver.customers c
join gizmobox_catalog_noori.silver.addresses a on a.customer_id = c.customer_id
```
