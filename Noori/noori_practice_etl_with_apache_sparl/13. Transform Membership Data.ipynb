## Transform memberships data
### 1. Extract customerid from the filepath
### 2. Write transformed data to the silver schema

```sql
select 
regexp_extract(path, r"/(\d+)\.png$", 1) as customer_id,
content as membership_card

from gizmobox_catalog_noori.bronze.v_memberships
```

```sql

create table if not exists gizmobox_catalog_noori.silver.memberships
select 
regexp_extract(path, r"/(\d+)\.png$", 1) as customer_id,
content as membership_card

from gizmobox_catalog_noori.bronze.v_memberships
```
