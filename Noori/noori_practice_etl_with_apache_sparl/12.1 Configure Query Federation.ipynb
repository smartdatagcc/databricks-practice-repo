## Configure access to azure ql database via lakehouse federation
### 1. create connection 
### 2. Create foreign catalog

```sql
CREATE CONNECTION asql_practice_db_conn_noori
    TYPE sqlserver
    OPTIONS (
       host 'gizmobox-practice-server.database.windows.net',
       port '1433',
       user 'gizmoboxadmin',
       password 'Databricks@123');
       
```

```sql
CREATE FOREIGN CATALOG IF NOT EXISTS asql_practice_db_conn_noori_catalog USING CONNECTION asql_practice_db_conn_noori
OPTIONS (database 'gizmobox-practice-db') 
```

```sql
show catalogs;
USE CATALOG practice_sql_db_connection_catalog;
show schemas;
use schema dbo;
select * from refunds
```

```sql
select * from practice_sql_db_connection_catalog.dbo.refunds r
join gizmobox_catalog_noori.silver.payments p on p.payment_id = r.payment_id
```
