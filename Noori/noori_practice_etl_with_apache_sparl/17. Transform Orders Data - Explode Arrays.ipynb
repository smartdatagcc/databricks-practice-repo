#### Transform Orders Data - Explode Arrays
##### 1. Access elements from the JSO Objects
##### 2. Deduplicate Array Elements
##### 3. Explode Arrays
##### 4. Write the Transformed Data to Silver Schema

```sql
select 
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
json_value.items

from gizmobox_catalog_noori.silver.orders_json
```

```sql


select 
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
array_distinct(json_value.items) as items

from gizmobox_catalog_noori.silver.orders_json
```

```sql

create or replace temporary view tv_orders_exploaded
as
select 
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
explode(array_distinct(json_value.items)) as items

from gizmobox_catalog_noori.silver.orders_json
```

```sql
select * from tv_orders_exploaded
```

```sql
select 
order_id,
order_status,
payment_method,
total_amount,
transaction_timestamp,
customer_id,
items.item_id,
items.name,
items.price,
items.quantity,
items.category,
items.details.brand,
items.details.color
from tv_orders_exploaded
```

```sql
create table if not exists gizmobox_catalog_noori.silver.orders
as

select 
order_id,
order_status,
payment_method,
total_amount,
transaction_timestamp,
customer_id,
items.item_id,
items.name,
items.price,
items.quantity,
items.category,
items.details.brand,
items.details.color
from tv_orders_exploaded
```

```sql
select * from gizmobox_catalog_noori.silver.orders
```
