# 05. Extract Addresses Data using Read_files Function - tsv

## Limitation of csv file format in select statement

```sql
%python
df = spark.read.csv("/Volumes/gizmobox_catalog_noori/landing/vol_operational_data/addresses/")
display(df)
```

```sql
%python
df = (spark.read.format("csv")
.option("header", "true")
.option("delimiter", "\t")
.load("/Volumes/gizmobox_catalog_noori/landing/vol_operational_data/addresses/")
)
display(df)
```

## 2. Use read_files functon to overcome the limitation

```sql
%python
df.writeTo("gizmobox_catalog_noori.bronze.py1_addresses").createOrReplace()
```
