Read file from Azure Sql via JDBC

```python
df = (
    spark.read.format('jdbc')
    .option('url', 'jdbc:sqlserver://gizmobox-practice-server.database.windows.net:1433;database=gizmobox-practice-db')
    .option('dbtable', 'Refunds')
    .option('user', 'gizmoboxadmin')
    .option('password', 'Databricks@123')
    .load()
)
display(df)
```

```python
df.writeTo("gizmobox_catalog_noori.bronze.py1_refunds").createOrReplace()
```
