### Join Customer and Address
##### join customer data with address data to create a customer_address table which contains the address of each customer on the same record

```python
df_customer = spark.read.table('gizmobox_catalog_noori.silver.py1_customers')
display(df_customer)

df_addresses = spark.read.table('gizmobox_catalog_noori.silver.py1_addresses')
display(df_addresses)
```

```python
df_Customer_addresses = (df_customer
                         .join(df_addresses, on='customer_id', how='inner')
                         .select(
                            df_customer.customer_id, 
                            df_customer.customer_name, 
                            df_customer.date_of_birth, 
                            df_customer.email, 
                            df_customer.member_since, 
                            df_customer.telephone, 
                            df_addresses.shipping_address_line_1, 
                            df_addresses.shipping_city, 
                            df_addresses.shipping_state, 
                            df_addresses.shipping_postcode, 
                            df_addresses.billing_address_line_1, 
                            df_addresses.billing_city, 
                            df_addresses.billing_state, 
                            df_addresses.billing_postcode 
                         )
)

# df_Customer_addresses = (
#     df_customer.join(df_addresses, df_customer.customer_id == df_addresses.customer_id, 'inner')
#     .select()
# )
display(df_Customer_addresses)
                         
```

```python
df_Customer_addresses.writeTo('gizmobox_catalog_noori.gold.py1_customer_addresses').createOrReplace()
```
