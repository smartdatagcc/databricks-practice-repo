## Transform memberships data
### 1. Extract customerid from the filepath
### 2. Write transformed data to the silver schema

```python
df=spark.read.table('gizmobox_catalog_noori.bronze.py1_memberships')
display(df)
```

```python
from pyspark.sql.functions import regexp_extract,col

df_formatted = df.select(
    regexp_extract('path', r"/(\d+)\.png$", 1).alias('customer_id'),
          col('content').alias('membership_card')
)
display(df_formatted)
```

```python
df_formatted.writeTo('gizmobox_catalog_noori.silver.py1_memberships').createOrReplace()
```
