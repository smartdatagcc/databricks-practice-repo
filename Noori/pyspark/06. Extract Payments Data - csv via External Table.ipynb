```python
%python
payment_schema =  'payment_id integer, order_id integer, payment_timestamp timestamp, palyment_status integer, payment_method string'

```

```python
%python
df = (
    spark.read.format("csv")
.option("delimiter", ",")
.schema(payment_schema)
.load('abfss://databrickspracticecontainer@databricksstgnoori.dfs.core.windows.net/Landing/external_data/payments')
)
display(df)
```

```python
%python
df.writeTo("gizmobox_catalog_noori.bronze.py1_payments").createOrReplace()
```

Creating schema with pyspark

```python
%python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, TimestampType
py_payment_scema = StructType ([
  StructField('payment_id', IntegerType(), True),
  StructField('order_id', IntegerType(), True),
  StructField('payment_timestamp', TimestampType(), True),
  StructField('payment_status', IntegerType(), True),
  StructField('payment_method', StringType(), True)
])
```

```python
%python
df = (
    spark.read.format("csv")
.option("delimiter", ",")
.schema(py_payment_scema)
.load('abfss://databrickspracticecontainer@databricksstgnoori.dfs.core.windows.net/Landing/external_data/payments')
)
display(df)
```

```python
%python
df.writeTo("gizmobox_catalog_noori.bronze.py11_payments").createOrReplace()
       
```
