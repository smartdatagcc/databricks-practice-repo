### 1. Difference between Create or REplace and Drop and Create Table statements

### Behaviour of the Drop and Create Statements

```sql
Drop table if exists gizmobox_catalog_noori.delta_lake.companies;
```

```sql

create or replace table gizmobox_catalog_noori.delta_lake.companies (
  company_name string,
  founded_date date,
  country string
)
comment 'This is a table of companies'
tblproperties ('sensitivity' = 'true', 'delta.enabledeletionvectors' = 'false');

insert into gizmobox_catalog_noori.delta_lake.companies
(company_name, founded_date, country)
values
('Microsoft', '2019-02-01', 'USA'),
('Google', '1996-01-01', 'USA'),
('Amazon', '1998-01-01', 'China');
```

```sql
desc history gizmobox_catalog_noori.delta_lake.companies
```

### Behaviour of the Create or Replace Statement

```sql
Drop table if exists gizmobox_catalog_noori.delta_lake.companies;
create table gizmobox_catalog_noori.delta_lake.companies (
compay_id bigint not null generated always as identity (start with 1 increment by 1),
  company_name string not null,
  founded_date date comment 'The date the company was founded',
  founded_year int generated always as (year(founded_date)),
  country string
)
comment 'This is a table of companies'
tblproperties ('sensitivity' = 'true', 'delta.enabledeletionvectors' = 'false');

insert into gizmobox_catalog_noori.delta_lake.companies
(company_name, founded_date, country)
values
('Microsoft', '2019-02-01', 'USA'),
('Google', '1996-01-01', 'USA'),
('Amazon', '1998-01-01', 'China');
```

```sql
desc history gizmobox_catalog_noori.delta_lake.companies
```

## 2. CTAS Statement
create table as select statement

```sql

create or replace table gizmobox_catalog_noori.delta_lake.companies_china AS

select * from gizmobox_catalog_noori.delta_lake.companies
where country='China'
```

```sql
select * from gizmobox_catalog_noori.delta_lake.companies_china
```

```sql
desc history gizmobox_catalog_noori.delta_lake.companies_china
```

```sql
desc extended gizmobox_catalog_noori.delta_lake.companies_china
```

```sql
drop table if exists gizmobox_catalog_noori.delta_lake.companies_china
```

```sql
create  table gizmobox_catalog_noori.delta_lake.companies_china AS

select cast(compay_id as int) as company_id, company_name, founded_date, country from gizmobox_catalog_noori.delta_lake.companies
where country='China'
```

```sql
alter table gizmobox_catalog_noori.delta_lake.companies_china
alter column founded_date comment 'The date the company was founded'
```

```sql
desc extended gizmobox_catalog_noori.delta_lake.companies_china
```

```sql
alter table gizmobox_catalog_noori.delta_lake.companies_china
alter column company_name set not null
```

```sql
desc extended gizmobox_catalog_noori.delta_lake.companies_china
```
