## Create Table - Table & Columbn Properties
Demonstrae adding table and column properites to the create table statement.

### 1. Table Properties
1.1 Comment = allows you to document the purpose of the table
1.2 TBLProperties - used to speify table level metadata or configuration settings

```sql
Drop table if exists gizmobox_catalog_noori.delta_lake_1.companies;
create table gizmobox_catalog_noori.delta_lake_1.companies (
  company_name string,
  founded_date date,
  country string
)
comment 'This is a table of companies'
tblproperties ('sensitivity' = 'true', 'delta.enabledeletionvectors' = 'false');
```

```sql
desc extended gizmobox_catalog_noori.delta_lake_1.companies
```

### 2. Column Properties
2.1 Not Null Constraints -  enforces data integrity and quality by ensuring that a specific column cannot contain null values
2.2 Comment -  documents the purpose or context of individual columns in a table

```sql
Drop table if exists gizmobox_catalog_noori.delta_lake_1.companies;
create table gizmobox_catalog_noori.delta_lake_1.companies (
  company_name string not null,
  founded_date date comment 'The date the company was founded',
  country string
)
comment 'This is a table of companies'
tblproperties ('sensitivity' = 'true', 'delta.enabledeletionvectors' = 'false');
```

```sql
desc extended gizmobox_catalog_noori.delta_lake_1.companies
```

### 2. Column Properties
2.3 Generated columns - derived or computed columns, whose values are computed at the time of inserting a new records
2.3.1. Generated Identity columns - used to generate an indentity for example a primary key value
2.3.2. Generated computed columns - automatically calculate and store derived values based on other columns in the same

Generated {Always | By Default } as Identity [([Start with start][increment by setp])]

```sql
Drop table if exists gizmobox_catalog_noori.delta_lake_1.companies;
create table gizmobox_catalog_noori.delta_lake_1.companies (
  company_id bigint not null generated always as identity (start with 1 increment by 1),
  company_name string not null,
  founded_date date comment 'The date the company was founded',
  country string
)
comment 'This is a table of companies'
tblproperties ('sensitivity' = 'true', 'delta.enabledeletionvectors' = 'false');
```

```sql
insert into gizmobox_catalog_noori.delta_lake_1.companies
(company_name, founded_date, country)
values
('Microsoft', '2019-02-01', 'USA'),
('Google', '1996-01-01', 'USA'),
('Amazon', '1998-01-01', 'USA');
```

```sql
select * from gizmobox_catalog_noori.delta_lake_1.companies
```
