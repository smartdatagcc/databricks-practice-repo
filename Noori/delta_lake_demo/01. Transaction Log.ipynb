## Understanding Delta Lake Transaction Log
Understand the cloud storage directory structure behind delta lake tables

##### 0. Create a newe schema under the demo catalog for this section of the course (delta_lake)

```sql
%sql
create schema if not exists gizmobox_catalog_noori.delta_lake_1
managed location 'abfss://databrickspracticecontainer@databricksstgnoori.dfs.core.windows.net/delta_lake_1'
```

##### 1. Create a Delta Lake Table

```sql
%sql
CREATE TABLE IF NOT EXISTS gizmobox_catalog_noori.delta_lake_1.companies
(
  company_name string,
  founded_date date,
  country string);
```

```sql
%sql
desc extended gizmobox_catalog_noori.delta_lake_1.companies
```
