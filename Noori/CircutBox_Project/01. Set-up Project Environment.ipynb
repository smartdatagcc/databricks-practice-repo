### 3. Create Schemas

###### 1. Schema Name : landing
###### Managed location: abfss://circutbox@databricksstgnoori.dfs.core.windows.net/landing
###### 
###### 1. Schema Name : lakehouse
###### Managed location: abfss://circutbox@databricksstgnoori.dfs.core.windows.net/lakehouse

```sql
use catalog circuitbox;

create schema if not exists landing
managed location 'abfss://circutbox@databricksstgnoori.dfs.core.windows.net/landing';

create schema if not exists lakehouse
managed location 'abfss://circutbox@databricksstgnoori.dfs.core.windows.net/lakehouse';

```

```sql
show schemas
```

### 4. Create Volume

###### Column Name: Operational_data
###### ADLS Path: abfss:/circuitbox@databricksstgnoori.dfs.core.windows.net/landing/opertional_data/
