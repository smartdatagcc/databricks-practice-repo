**Create External Table**

```sql
%python
from pyspark.sql.types import StructType,StructField,StringType,IntegerType,FloatType,TimestampType

payment_Struct=StructType([
StructField('paymentId',IntegerType(),True),
StructField('orderId',IntegerType(),True),
StructField('paymetOn',TimestampType(),True),
StructField('paymentStatus',IntegerType(),True),
StructField('paymentMethod',StringType(),True)
])
payment_schema='paymentId INTEGER,orderId INTEGER,paymetOn TIMESTAMP,paymentStatus INTEGER,paymentMethod STRING';

dfPayments=spark.read.format('csv').option('header','true').schema(payment_Struct).load('abfss://gizmobox-sankara@ourdbpracticeextdl.dfs.core.windows.net/landing/external_data/payments')
display(dfPayments)
```

```sql
%python
dfPayments.writeTo('gizmobox_sankara.bronze.py_payments').createOrReplace()
```
