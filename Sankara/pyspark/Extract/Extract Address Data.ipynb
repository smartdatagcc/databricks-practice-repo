**Extract Addresses Data From Files**

**1.Limitations Of CSV File Format In SELECT Statement**

```sql
%python
df = spark.read.csv("/Volumes/gizmobox_sankara/landing/operational_volume/addresses/")
display(df)
```

**2.Use read_file Function To Overcome Limitations**

```sql
%python
dfAddresses=spark.read.format("csv").option("header", "true").option("delimiter","\t").load("/Volumes/gizmobox_sankara/landing/operational_volume/addresses")
display(dfAddresses)
```

```sql
%python
dfAddresses.writeTo("gizmobox_sankara.bronze.py_addresses").createOrReplace();
```

```sql
%python
dfAddresses = spark.table("gizmobox_sankara.bronze.py_addresses")
display(dfAddresses)
```
