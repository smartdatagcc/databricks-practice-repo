**TransForm Membershps Data**

```sql
%python
df_memberShipsData=spark.table("gizmobox_gr.bronze.py_membership")
display(df_memberShipsData)
```

```sql
%python
from pyspark.sql import functions as f
df_memberShipsData_customer=df_memberShipsData.select(
    f.regexp_extract(df_memberShipsData.path, '.*/([0-9]+)\\.png$', 1).alias("customer_id"), 
    df_memberShipsData.content,
    df_memberShipsData.modificationTime
)
display(df_memberShipsData_customer)
```

**Write Transformed Data To Silver Schema**

```sql
%python
df_memberShipsData_customer.writeTo("gizmobox_gr.silver.py_membership").createOrReplace()
display(spark.table("gizmobox_gr.silver.py_membership"))
```
