**Join Customer Address**

```python
dfCustomers=spark.table("gizmobox_gr.silver.py_customers")
dfAddresses=spark.table("gizmobox_gr.silver.py_addresses")
display(dfCustomers)
display(dfAddresses)
```

```python
dfCustomerAddresses=(dfCustomers
                     .join(dfAddresses,[dfCustomers.customer_id==dfAddresses.customer_id],"inner")
                     .select(dfCustomers.customer_id.alias("customer_id"),
                             dfCustomers.customer_name.alias("customer_name"),
                             dfCustomers.date_of_birth.alias("date_of_birth"),
                             dfCustomers.email.alias("email"),
                             dfCustomers.member_since.alias("member_since"),
                             dfCustomers.telephone.alias("telephone"),
                             dfAddresses.shipping_address_line_1.alias("shipping_address_line_1"),
                             dfAddresses.shipping_city.alias("shipping_city"),
                             dfAddresses.shipping_state.alias("shipping_state"),
                             dfAddresses.shipping_postcode.alias("shipping_postcode"),
                             dfAddresses.billing_address_line_1.alias("billing_address_line_1"),
                             dfAddresses.billing_city.alias("billing_city"),
                             dfAddresses.billing_state.alias("billing_state"),
                             dfAddresses.billing_postcode.alias("billing_postcode")
                     ) )
display(dfCustomerAddresses)
```

**Write Customer Addresses Data To Gold Schema**

```python
dfCustomerAddresses.writeTo("gizmobox_gr.gold.py_customer_addresses").createOrReplace()
display(spark.table("gizmobox_gr.gold.py_customer_addresses"))
```
