# Remove the records with Null Customer_id

```python
df_customer = spark.read.table("gizmobox_sankara.bronze.py_customers")
display(df_customer)
```

```python
df_customer_filter = df_customer.filter(df_customer.customer_id.isNotNull())
display(df_customer_filter)
```

```python
%sql
SELECT *
FROM gizmobox_sankara.bronze.v_customers
WHERE customer_id IS NOT NULL
ORDER BY customer_id ASC
LIMIT 100
```

```python
df_customer_distinct = df_customer_filter.distinct()
display(df_customer_distinct)
```

# Remove the Duplicates

```python
from pyspark.sql import functions as f
df_max = df_customer_distinct.groupBy('customer_id').agg(f.max('created_timestamp'))
display(df_max)
```

```python
df_distinct_customers_timestamp = (
    df_customer_distinct.join(df_max, (df_customer_distinct.customer_id == df_max.customer_id) & (df_customer_distinct.created_timestamp == df_max['max(created_timestamp)'])).select(df_customer_distinct['*'])
)
display(df_distinct_customers_timestamp)
```

```python
%sql
CREATE OR REPLACE TEMPORARY VIEW v_customers_distinct 
AS
SELECT DISTINCT *
 FROM gizmobox_sankara.bronze.v_customers
WHERE customer_id IS NOT NULL
ORDER BY customer_id;
```

```python
%sql
SELECT customer_id,
      MAX(created_timestamp) AS max_created_timestamp
      FROM v_customers_distinct
      GROUP BY customer_id
```

```python
%sql
with cte_max AS
(
  SELECT customer_id,
      MAX(created_timestamp) AS max_created_timestamp
      FROM v_customers_distinct
      GROUP BY customer_id
)
SELECT t.*
    FROM v_customers_distinct t
    INNER JOIN cte_max m
    ON t.customer_id = m.customer_id
    AND t.created_timestamp = m.max_created_timestamp
    
```

```python
%sql
WITH cte_max AS
(
  SELECT customer_id,
      MAX(created_timestamp) AS max_created_timestamp
      FROM v_customers_distinct
      GROUP BY customer_id
)
SELECT CAST(t.created_timestamp AS TIMESTAMP) As created_timestamp,
    t.customer_id,
    t.customer_name,
    CAST(t.date_of_birth AS DATE) AS date_of_birth,
    t.email,
    CAST(t.member_since AS DATE) AS member_since,
    t.telephone
    FROM v_customers_distinct t
    INNER JOIN cte_max m
    ON t.customer_id = m.customer_id
    AND t.created_timestamp = m.max_created_timestamp
```

```python
%sql
DROP TABLE gizmobox_nara.silver.customers;
CREATE TABLE gizmobox_nara.silver.customers AS
WITH cte_max AS
(
  SELECT customer_id,
      MAX(created_timestamp) AS max_created_timestamp
      FROM v_customers_distinct
      GROUP BY customer_id
)
SELECT CAST(t.created_timestamp AS TIMESTAMP) As created_timestamp,
    t.customer_id,
    t.customer_name,
    CAST(t.date_of_birth AS DATE) AS date_of_birth,
    t.email,
    CAST(t.member_since AS DATE) AS member_since,
    t.telephone
    FROM v_customers_distinct t
    INNER JOIN cte_max m
    ON t.customer_id = m.customer_id
    AND t.created_timestamp = m.max_created_timestamp
```
