# CREATING EXTERNAL TABLE ON DATA


```python
df = (
    spark.read.format("csv")
    .option("delimiter",",")
    .schema(payment_schema)
    .load('abfss://gizmobox-nara@ourdbpracticeextdl.dfs.core.windows.net/Landing/External_data/Payments')
)
display(df)
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, TimestampType

py_payment_schema = StructType([
    StructField('paymentid', IntegerType()),
    StructField('orderid', IntegerType()),
    StructField('paymenttimestamp', TimestampType()),
    StructField('payment_status', IntegerType()),
    StructField('payment_method', StringType()) 
])
```

```python
df = (
    spark.read.format('csv')
      .option('delimiter', ',')
      .schema(py_payment_schema)
       .load('abfss://gizmobox-nara@ourdbpracticeextdl.dfs.core.windows.net/Landing/External_data/Payments')
       )
display(df)
```

```python
df.writeTo("gizmobox_nara.bronze.py1payments").createOrReplace()
```
