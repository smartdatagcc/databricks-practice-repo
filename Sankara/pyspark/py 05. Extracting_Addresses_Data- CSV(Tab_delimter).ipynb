### 1.LIMIATION Of SELECT Statement in CSV_FILES

```python
df = spark.read.csv("/Volumes/gizmobox_/landing/operational_volume/addresses/")
display(df)
```

```python
df_format = (
    spark.read.format('csv')
    .option("header", "true")
    .option("delimiter","\t")
    .load("/Volumes/gizmobox_nara/landing/operational_volume/addresses/")
    )
```

# 2.USING TABLED VALUED FUNCTION

# CREATE A VIEW FOR ADRESSES IN BRONZE LAYER

```python
df_format.writeTo("gizmobox_nara.bronze.py1addresses").createOrReplace()
```

```python
df = spark.table("gizmobox_nara.bronze.py1addresses")
display(df)
```
