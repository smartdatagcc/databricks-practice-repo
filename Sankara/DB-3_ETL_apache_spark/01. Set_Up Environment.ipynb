# CREATE _VOLUME_

```python
%sql
USE CATALOG gizmobox_sankara;

-- Drop conflicting volume in default schema that occupies the same storage location
DROP VOLUME IF EXISTS gizmobox_sankara.default.operational_data;

USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_volume
 LOCATION 'abfss://gizmobox-sankara@ourdbpracticeextdl.dfs.core.windows.net/landing/operational_data/'
COMMENT 'operational data volume';

```
