### Transform Memberships Data

    1. Extract Customer_Id from the file path
    2. Write Transformed data to the Silver schema

```python
dfMembershipsData = spark.table("gizmobox_sivan.bronze.v_memberships")
display(dfMembershipsData)
```

### 1. Extract Customer_ID from the file path
[regexp_extract Doc](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/regexp_extract)

[sql-ref-regexp-pattern](https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/sql-ref-regexp-pattern)

```python
from pyspark.sql import functions as f
dfMembershipsFinalData = (
    dfMembershipsData.select(
        f.regexp_extract(dfMembershipsData.path, '.*/([0-9]+)\\.png$',1).alias("customer_id"),
        dfMembershipsData.content.alias("membership_card")
    )
)

display(dfMembershipsFinalData)
```

###2.Write Transformed data to the Silver schema

```python
dfMembershipsFinalData.writeTo("gizmobox_sivan.silver.py_memberships").createOrReplace()
```

```python
df = spark.table("gizmobox_sivan.silver.py_memberships")
display(df)
```
