### Extract Data from the Orders JSON File
 1. Query Orders File Using JSON Format
 2. Query Orders File using TEXT Format
 3. Create Orders View in Bronze Schema

### 1. Query Orders File using JSON Format

```python
df = spark.read.format("json").load("/Volumes/gizmobox_sivan/landing/operational_data_volume/orders")
display(df);
```

### 2. Query Orders File using TEXT Format

```python
df = spark.read.text("/Volumes/gizmobox_sivan/landing/operational_data_volume/orders")
display(df);
```

### 3. Craeted Orders view in Bronze `schema`

```python
df.writeTo("gizmobox_sivan.bronze.py_orders").createOrReplace()
```

```python
%sql
select * from gizmobox_sivan.bronze.py_orders;
```
