### Monthly Order Summary

For each of the customer, produce the following summary per month
 1. total oders
 2. total items bought
 3. total amount spent

```python
dfOrders = spark.table("gizmobox_sivan.silver.orders")
dfOrders.display()
```

```python
from pyspark.sql import functions as f
dfOrder_monthly_Summary = (
    dfOrders
        .groupBy(["order_month","customer_id"])
        .agg(
            f.countDistinct("order_id").alias("total_orders"),
            f.sum("quantity").alias("total_items"),
            f.sum(f.col("price") * f.col("quantity")).alias("total_amount")
        )
        .select(
            f.date_format("","")
        )
)
```

```python
%sql
SELECT
    date_format(transaction_timestamp, 'yyyy-MM') as order_month,
    customer_id,
    count(distinct order_id) as total_orders,
    sum(quantity) as total_items,
    sum(price* quantity) as total_amount
from gizmobox_sivan.silver.orders
group by order_month, customer_id;
```

```python
%sql
CREATE TABLE IF NOT EXISTS gizmobox_sivan.gold.order_summary_monthly
AS
SELECT
    date_format(transaction_timestamp, 'yyyy-MM') as order_month,
    customer_id,
    count(distinct order_id) as total_orders,
    sum(quantity) as total_items,
    sum(price* quantity) as total_amount
from gizmobox_sivan.silver.orders
group by order_month, customer_id;
```

```python
%sql
select * from  gizmobox_sivan.gold.order_summary_monthly;
```
