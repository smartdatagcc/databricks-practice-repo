### Query Orders data as JSON strings

1. Extract top level column values
2. Extract Array elements
3. Extract Nested column values
4. CAST Column values to a specific Data Type

```sql
select * from gizmobox_sivan.bronze.v_orders
```

###1. Extract Top level object Values


Top level object  column_name:extraction_path

```sql
select
    value:order_id,
    value
from gizmobox_sivan.bronze.v_orders;
```

###2. Extract Array elements

To extract array elements - use the indices in brackets column_name:extraction_path [index]

```sql
select    
    value:items[0] as item1,
    value:items[1] as item2,
    value:items,
    value
from gizmobox_sivan.bronze.v_orders;
```

### 3. Extract Nested Column values

To extract nested fields, use dot notation column_name:extraction_path[index].leaf_node

```sql
select    
    value:items[0].item_id as item_id,
    value:items[0] as item1,
    value:items[1] as item2,
    value:items,
    value
from gizmobox_sivan.bronze.v_orders;
```

### 4. CAST Column values to a specific Data Type

```sql
select    
    value:items[0].item_id::int as item_id,
    value:items[0] as item1,
    value:items[1] as item2,
    value:items,
    value
from gizmobox_sivan.bronze.v_orders;
```
