### 1. Query Single JSON File

```python
%fs ls /Volumes/gizmobox_sivan/landing/operational_data_volume/customers
```

```python
df = spark.read.format("json").load("/Volumes/gizmobox_sivan/landing/operational_data_volume/customers/customers_2024_10.json")
display(df)
```

### 2. Query Multiple JSON Files

```python
df = spark.read.format("json").load("/Volumes/gizmobox_sivan/landing/operational_data_volume/customers/customers*.json")
display(df)
```

### 3. Query all the files in a folder

```python
df = spark.read.format("json").load("/Volumes/gizmobox_sivan/landing/operational_data_volume/customers")
display(df)
```

```python
df = spark.read.json("/Volumes/gizmobox_sivan/landing/operational_data_volume/customers")
display(df)
```

### 4. Select File Meta Data

```python
df_metadata = df.select("_metadata.file_path","*")
display(df_metadata);
```

### 5. Create a Table in Bronze schema

```python
df_metadata.write.format("delta").mode("overwrite").saveAsTable("gizmobox_sivan.bronze.py_customers")
```

```python
df_metadata.writeTo("gizmobox_sivan.bronze.py_customers").createOrReplace
```

```python
df = spark.table("gizmobox_sivan.bronze.py_customers")
display(df)
```
