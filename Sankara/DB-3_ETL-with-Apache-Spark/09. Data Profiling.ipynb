###1. Profile Data using UI

```sql
select * from gizmobox_sivan.bronze.v_customers
```

###2. Profile Data using DBUTILS Package (dbutils.data.summarize method)

```sql
%python
df = spark.table('gizmobox_sivan.bronze.v_customers')
dbutils.data.summarize(df)
```

###3.Profile Data manually

**3.1 Count Function**

```sql
select count(*), count(customer_id), count(email), count(telephone)
from gizmobox_sivan.bronze.v_customers
```

**3.2 COUNT_IF Function**

```sql
select count_if(customer_id is null), count_if(email is null), count_if(telephone is null)
from gizmobox_sivan.bronze.v_customers;
```

**3.3 WHERE Clause**

```sql
select count(*)
from gizmobox_sivan.bronze.v_customers
where customer_id is null;
```

**3.4 DISTINCT Keyword**

```sql
select 
count(distinct customer_id), count(distinct email), count(distinct telephone)
from gizmobox_sivan.bronze.v_customers
where customer_id is not null
```
