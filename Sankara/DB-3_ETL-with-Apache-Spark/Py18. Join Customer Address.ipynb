### Join Customer and Address

Join Customer data with address data to create customer_address table which contains the addresses of each customer on the same record

```python
dfCustomers = spark.table("gizmobox_sivan.silver.py_customers")
dfCustomers.display()
```

```python
dfAddresses = spark.table("gizmobox_sivan.silver.py_addresses")
dfAddresses.display()
```

```python
from pyspark.sql import functions as f
dfCustomerAddresses = (
    dfCustomers
        #.join(dfAddresses, on='customer_id', how='inner')
        #.join(dfAddresses, [dfCustomers.customer_id == dfAddresses.customer_id], "inner")
        .join(dfAddresses, "customer_id","inner")
        .select(
                dfCustomers.customer_id,
                dfCustomers.customer_name,
                dfCustomers.email,
                dfCustomers.date_of_birth,
                dfCustomers.member_since,
                dfCustomers.telephone,
                dfAddresses.shipping_address_line_1,
                dfAddresses.shipping_city,
                dfAddresses.shipping_state,
                dfAddresses.shipping_postcode,
                dfAddresses.billing_address_line_1,
                dfAddresses.billing_city,
                dfAddresses.billing_state,
                dfAddresses.billing_postcode
        )
)

dfCustomerAddresses.display()
```

```python
dfCustomerAddresses.writeTo("gizmobox_sivan.gold.py_customer_address").createOrReplace()
```
