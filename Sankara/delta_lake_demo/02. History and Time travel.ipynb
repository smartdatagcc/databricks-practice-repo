## History and Time travel
##### 1. Query delta lake table history
##### 2. Query privious versions ofthe data
##### 3. Query data from a specific time
##### 4. Restore data to a speciic version


##### 1. Query delta lake table history

```sql
describe History gizmobox_sankara.delta_lake.companies
```

##### 2. Query privious versions ofthe data

```sql
select * from gizmobox_sankara.delta_lake.companies
version as of 1
```

##### 3. Query data from a specific time

```sql
select * from gizmobox_sankara.delta_lake.companies
TIMESTAMP AS OF '2026-08-17T16:46:30.000+00:00'
```

##### 4. Restore data to a speciic version

```sql
RESTORE TABLE gizmobox_sankara.delta_lake.companies
version as of 1
```

```sql
select * from gizmobox_sankara.delta_lake.companies
```

```sql
describe history gizmobox_sankara.delta_lake.companies
```
