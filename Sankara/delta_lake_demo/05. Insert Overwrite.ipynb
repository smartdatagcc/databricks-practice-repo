## Insert Overwrite
1. REplace all the data in a table
2. Replace all the data from a specific partition
3. How to handle schema changes

##### Insert Overwrite -  overwrites the existing data in a table or a speific partition with the new data.
##### Insert into - Appends new data

### 1. Replace all the data in a table 

```sql
drop table if exists gizmobox_sankara.delta_lake.gold_companies
```

```sql
create table gizmobox_sankara.delta_lake.gold_companies (
    
  company_name string ,
  founded_date date ,
  country string
);


insert into gizmobox_sankara.delta_lake.gold_companies
(company_name, founded_date, country)
values

('Google', '1996-01-01', 'USA'),
('Amazon', '1998-01-01', 'China');

select * from gizmobox_sankara.delta_lake.gold_companies
```

```sql


Drop table if exists gizmobox_sankara.delta_lake.bronze_companies;
create table gizmobox_sankara.delta_lake.bronze_companies (
    company_name string ,
  founded_date date ,
  country string
);


insert into gizmobox_sankara.delta_lake.bronze_companies
(company_name, founded_date, country)
values
('Microsoft', '2019-02-01', 'USA'),
('Google', '1996-01-01', 'DUBAI'),
('Apple', '1986-01-01', 'INDIA'),
('Tencent', '1994-01-01', 'UK'),
('Amazon', '1998-01-01', 'SRILANKA');

select * from gizmobox_sankara.delta_lake.bronze_companies
```

```sql
insert overwrite table gizmobox_sankara.delta_lake.gold_companies
select * from gizmobox_sankara.delta_lake.bronze_companies
```

```sql
select * from gizmobox_sankara.delta_lake.gold_companies
```

```sql
desc history gizmobox_sankara.delta_lake.gold_companies
```

### 2. Replace all the data from specific partition

```sql
create table gizmobox_sankara.delta_lake.gold_companies_partitioned_NEW (
    
  company_name string ,
  founded_date date ,
  country string
)
partitioned by (country);


insert into gizmobox_sankara.delta_lake.gold_companies_partitioned
(company_name, founded_date, country)
values

('Google', '1996-01-01', 'USA'),
('Amazon', '1998-01-01', 'China');

select * from gizmobox_sankara.delta_lake.gold_companies_partitioned
```

```sql
desc extended gizmobox_sankara.delta_lake.gold_companies_partitioned_NEW
```

```sql
Drop table if exists gizmobox_sankara.delta_lake.bronze_companies_USA;
create table gizmobox_sankara.delta_lake.bronze_companies_USA (
    company_name string ,
  founded_date date ,
  country string
);


insert into gizmobox_sankara.delta_lake.bronze_companies_USA
(company_name, founded_date, country)
values
('Microsoft', '2019-02-01', 'USA'),
('Google', '1996-01-01', 'USA'),
('Apple', '1986-01-01', 'USA'),
('Tencent', '1994-01-01', 'USA');

select * from gizmobox_sankara.delta_lake.bronze_companies_USA
```

```sql
insert overwrite table gizmobox_sankara.delta_lake.gold_companies_partitioned
partition (country='USA')
select company_name,
        founded_date 
from gizmobox_sankara.delta_lake.bronze_companies_USA
```

```sql
select * from gizmobox_sankara.delta_lake.gold_companies_partitioned
```

```sql
desc history gizmobox_sankara.delta_lake.gold_companies_partitioned
```

### 3.How to handle schema changes

insert overwrite -> Use to overite the data in a table or a partition when there are no schema changes.

Create or replace table -> use when there are schema changes.
