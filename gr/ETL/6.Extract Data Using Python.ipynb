**Extract Data-Using PySpark**
- Run Sql Command Using Python-spark.sql function
- Spark Data Frame Reader API
- Read Tables Using spark.table Functions

**Run Sql Command Using Python-spark.sql function**

```sql
%python
df=spark.sql("SELECT * FROM json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers`");
display(df);
```

```sql
%python
df=spark.sql('''CREATE OR REPLACE TEMPORARY VIEW tv_customer_info 
             AS 
             SELECT * FROM json.`/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers`'''
             );
```

```sql
SELECT * FROM tv_customer_info
```

**Spark Data Frame Reader API**

```sql
%python
df=spark.read.format("json").load("/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers")
display(df)
```

```sql
%python
spark.read.format("json").load("/Volumes/gizmobox_catalog_subbu/landing/operational_data_volume/customers").createOrReplaceTempView("tv_customers_info")
```

```sql
SELECT * FROM tv_customers_info
```
