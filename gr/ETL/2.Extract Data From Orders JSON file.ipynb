**Extract Data From Orders JSON file**

**1.Query To Read Orders Json File**

```sql
SELECT * FROM JSON.`/Volumes/gizmobox_gr/landing/operational_data/orders`
```

**2.Query To Read Orders Json File Data To Text Format**

```sql
SELECT * FROM Text.`/Volumes/gizmobox_gr/landing/operational_data/orders`
```

**3.Create a View for Bronze Schema**

```sql
CREATE OR REPLACE VIEW gizmobox_gr.bronze.v_orders AS
SELECT *,_metadata.file_path FROM Text.`/Volumes/gizmobox_gr/landing/operational_data/orders`
```
