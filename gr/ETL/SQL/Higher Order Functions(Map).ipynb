**Higher Order Functions(Map)**
- TRANSFORM_KEYS
- TRANSFORM_VALUES
- MAP_FILTER

```sql
select json_value,json_value.customer_id, explode(array_distinct(json_value.items)) 
from gizmobox_gr.silver.orders_json
```

**Transform Keys**

```sql
select
transform_keys(map(json_value.items.details.brand[0], json_value.items.details.color[0]), (key, value) -> UPPER(key)) as items
from gizmobox_gr.silver.orders_json
```

**Filter**

```sql
select explode(array_distinct(filter(json_value.items, x -> x.name like'%Gaming%' and x.details.brand like'%GoPro%'))) GoPro
from gizmobox_gr.silver.orders_json;
```

**Exists**

```sql
select json_value.items,exists(json_value.items, x -> x.details.brand ='GoPro') GoPro
from gizmobox_gr.silver.orders_json;
```

**Aggregate**

```sql
select json_value.items, aggregate(json_value.items, 0L,(acc,x)->acc+x.price) as total_price
from gizmobox_gr.silver.orders_json;
```

**Array with more than one object**

```sql
select json_value.items,
transform(json_value.items,x->named_struct(
    'name',UPPER(x.name),
    'price',x.price*10)) transformeddata
from gizmobox_gr.silver.orders_json
where size(json_value.items) > 1;
```
