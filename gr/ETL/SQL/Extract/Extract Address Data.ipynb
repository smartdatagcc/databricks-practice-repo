**Extract Addresses Data From Files**

**1.Limitations Of CSV File Format iIn SELECT Statement**

```sql
SELECT * FROM csv.`/Volumes/gizmobox_gr/landing/operational_data/addresses`
```

**2.Use read_file Function To Overcome Limitations**

```sql
SELECT * FROM read_files('/Volumes/gizmobox_gr/landing/operational_data/addresses', format=>'csv', delimiter=>'\t', header=>'true')
```

**3.Create View in Bronze Schema**

```sql
CREATE OR REPLACE VIEW gizmobox_gr.bronze.v_addresses AS 
SELECT * FROM read_files('/Volumes/gizmobox_gr/landing/operational_data/addresses', format=>'csv', delimiter=>'\t', header=>'true')
```

```sql
SELECT * FROM v_addresses
```
