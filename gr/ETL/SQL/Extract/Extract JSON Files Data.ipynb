**Extract Data From The JSON file**
- Query SingleFIle
- Query List Of All Files Using Wildcard Characters
- Query All The Files In The Folders
- Select File Meta Data
- Register Files In Unity Catalogs Using Views
- Create Temporary View
- Create Global View

**1. Query Single Json File**

```sql
SELECT * FROM JSON.`/Volumes/gizmobox_gr/landing/operational_data/customers/customers_2024_10.json`
```

**2.Query Multiple Json Files**

```sql
SELECT * FROM JSON.`/Volumes/gizmobox_gr/landing/operational_data/customers/customers_2024_*.json`
```

**3.Query All The Json Files From Folder**

```sql
SELECT * FROM JSON.`/Volumes/gizmobox_gr/landing/operational_data/customers`
```

**4.Select File Meta Data**

```sql
SELECT input_file_name() As FilePath,--Depricated from Databricks Runtime 13.3 LTS
_metadata As FileMetaData,
_metadata.file_path As FilePathName
FROM JSON.`/Volumes/gizmobox_gr/landing/operational_data/customers/customers_2024_*.json`
```

**5.Registering Files In Unity Catalog Using Views**

```sql
CREATE OR REPLACE VIEW gizmobox_gr.bronze.v_customers AS
SELECT *,_metadata.file_path as FilePath FROM JSON.`/Volumes/gizmobox_gr/landing/operational_data/customers`
```

```sql
SELECT * FROM gizmobox_gr.bronze.v_customers
```
