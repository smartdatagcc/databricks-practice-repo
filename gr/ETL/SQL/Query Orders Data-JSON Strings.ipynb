**Query Orders Data-JSON Strings**

```sql
SELECT 
value:order_id AS order_id,
value:order_date AS order_date,
value:order_customer_id AS order_customer_id,
value:order_status AS order_status,
value:items AS items,
value,
file_path
 FROM gizmobox_gr.bronze.v_orders;
```

```sql
SELECT 
value:order_id::int AS order_id,
value:order_date::date AS order_date,
value:order_customer_id::int AS order_customer_id,
value:order_status AS order_status,
value:items AS items,
value:items[0].item_id AS item_id,
value:items[1].name AS item_name,
value:items[2].price AS item_price,
value:items[3].quantity AS item_quantity,
value,
file_path
 FROM gizmobox_gr.bronze.v_orders;
```

**CAST Values To Specific Types**

```sql
SELECT 
value:order_id::int AS order_id,
value:order_date::date AS order_date,
value:order_customer_id::int AS order_customer_id,
value:order_status AS order_status,
value:items AS items,
value:order_items AS order_items,
value,
file_path
 FROM gizmobox_gr.bronze.v_orders;
```
