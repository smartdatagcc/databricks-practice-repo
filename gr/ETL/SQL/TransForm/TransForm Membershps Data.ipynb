**TransForm Membershps Data**

```sql
SELECT regexp_extract(path, '.*/([0-9]+)\\.png$', 1) AS CustomerId , content as membership_card 
FROM gizmobox_gr.bronze.v_membership;
```

```sql
CREATE TABLE gizmobox_gr.silver.memberships
SELECT regexp_extract(path, '.*/([0-9]+)\\.png$', 1) AS CustomerId , content as membership_card 
FROM gizmobox_gr.bronze.v_membership;
```
