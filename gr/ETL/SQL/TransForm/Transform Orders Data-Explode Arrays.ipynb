**Access Elements From JSON Object**

```sql
SELECT json_value.customer_id,
json_value.transaction_timestamp,
json_value.order_id,
json_value.order_date,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
explode(array_distinct(json_value.items)) item
--explode(array_distinct(json_value.items.details)) itemDetails
FROm gizmobox_gr.silver.orders_json
```

```sql
CREATE OR REPLACE TEMPORARY VIEW v_orders_json_explode AS
SELECT json_value.customer_id,
json_value.transaction_timestamp,
json_value.order_id,
json_value.order_date,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
explode(array_distinct(json_value.items)) item,
explode(array_distinct(json_value.items.details)) itemDetails
FROm gizmobox_gr.silver.orders_json
```

```sql
CREATE OR REPLACE TABLE gizmobox_gr.silver.orders
SELECT 
customer_id,
transaction_timestamp,
order_id,
order_date,
order_status,
payment_method,
total_amount,
item.item_id,
itemDetails.brand,
itemDetails.color,
item.name,
item.price,
item.quantity
 FROM v_orders_json_explode;
```
