**3.Profile Data Manually**
- COUNT
- COUNT_IF
- MIN
- MAX
- WHERE Clause

```sql
SELECT COUNT(*),COUNT(email),COUNT(customer_id),COUNT(DISTINCT email),COUNT(DISTINCT customer_id) 
FROM gizmobox_gr.bronze.v_customers;

SELECT COUNT(*),COUNT(email),COUNT(customer_id),COUNT(DISTINCT email),COUNT(DISTINCT customer_id) 
FROM gizmobox_gr.bronze.v_customers
WHERE customer_id IS NOT NULL;
```
