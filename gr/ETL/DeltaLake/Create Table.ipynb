**Identity**

```python
%sql
drop table gizmobox_gr.delta_lake.companies;
create table if not exists gizmobox_gr.delta_lake.companies
(
    company_id bigint not null generated always as identity(start with 1 increment by 1),
    company_name string not null,
    founded_date date comment 'This is founded date',
    country string)
comment 'This is a table for companies'
tblproperties ('sensitivity' = 'true',
  'delta.enabledeletionvectors' = 'false');
```

```python
%sql
insert into gizmobox_gr.delta_lake.companies(company_name,founded_date,country)
values
('Mac','2019-01-01', 'USA');
```

```python
%sql
drop table gizmobox_gr.delta_lake.companies;
create table if not exists gizmobox_gr.delta_lake.companies
(
    company_id bigint not null generated always as identity(start with 1 increment by 1),
    company_name string not null,
    founded_date date comment 'This is founded date',
    founded_year int generated always as (year(founded_date)),
    country string)
comment 'This is a table for companies'
tblproperties ('sensitivity' = 'true',
  'delta.enabledeletionvectors' = 'false');
```
