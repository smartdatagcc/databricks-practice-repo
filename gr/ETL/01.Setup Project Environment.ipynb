**2. CREATE EXTERNAL LOCATION IF NOT EXISTS**

```sql
CREATE EXTERNAL LOCATION if NOT EXISTS databricks_practice_ext_dl_gizmoboxgr
URL 'abfss://gizmoboxgr@ourdbpracticeextdl.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-practice-ext-sc`)
COMMENT 'External location for databricks_practice_ext_dl_gizmoboxgr';
```

**3. CREATE CATALOG gizmobox**

```sql
show catalogs
```

**4. CREATE CATALOG**

```sql
CREATE CATALOG IF NOT EXISTS gizmobox_gr
      MANAGED LOCATION 'abfss://gizmoboxgr@ourdbpracticeextdl.dfs.core.windows.net/'
      COMMENT 'Data Lake house gizmobox catalog'
```

**5. CREATE SCHEMAS**

- Landing
- Bronze
- Silver
- Gold

```sql
select current_catalog();
use catalog gizmobox_gr;
```

```sql
use catalog gizmobox_gr;

CREATE SCHEMA IF NOT EXISTS landing
MANAGED LOCATION 'abfss://gizmoboxgr@ourdbpracticeextdl.dfs.core.windows.net/landing'
COMMENT 'landing schema for gizmobox' ;

CREATE SCHEMA IF NOT EXISTS bronze
MANAGED LOCATION 'abfss://gizmoboxgr@ourdbpracticeextdl.dfs.core.windows.net/bronze'
COMMENT 'bronze schema for gizmobox' ;

CREATE SCHEMA IF NOT EXISTS silver
MANAGED LOCATION 'abfss://gizmoboxgr@ourdbpracticeextdl.dfs.core.windows.net/silver'
COMMENT 'silver schema for gizmobox' ;

CREATE SCHEMA IF NOT EXISTS gold
MANAGED LOCATION 'abfss://gizmoboxgr@ourdbpracticeextdl.dfs.core.windows.net/gold'
COMMENT 'gold schema for gizmobox' 
```

**6. CREATE VOLUME**

```sql
USE CATALOG gizmobox_gr;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://gizmoboxgr@ourdbpracticeextdl.dfs.core.windows.net/landing/operational_data'
    COMMENT 'volume for operational data'
```
