**TransForm Addresses Data**

```sql
%python
df_addresses=spark.table("gizmobox_gr.bronze.py_addresses")
display(df_addresses)
```

**1.Create One Record For Each Customer With  Both Addresses**

```sql
%python
from pyspark.sql import functions as f
df_addresses_pivot=(df_addresses
                    .groupby('customer_id')
                    .pivot('address_type',['shipping','billing'])
                    .agg(
                        f.max('address_line_1').alias('address_line_1'),
                        f.max('city').alias('city'),
                        f.max('state').alias('state'),
                        f.max('postcode').alias('postcode')
                       )
                   )
display(df_addresses_pivot)
```

**Write Transformed Data To Silver Schema**

```sql
%python
df_addresses_pivot.writeTo("gizmobox_gr.silver.py_addresses").createOrReplace()
display(spark.table("gizmobox_gr.silver.py_addresses"))
```
