**Extract Data From Orders JSON file**

**1.Query To Read Orders Json File**

```sql
%python
dfOrders=spark.read.json("/Volumes/gizmobox_gr/landing/operational_data/orders")
display(dfOrders)
```

**2.Query To Read Orders Json File Data To Text Format**

```sql
%python
dfOrdersText=spark.read.text("/Volumes/gizmobox_gr/landing/operational_data/orders")
display(dfOrdersText)
```

**3.Create a View for Bronze Schema**

```sql
%python
dfOrdersText.writeTo("gizmobox_gr.bronze.py_orders").createOrReplace()
```
