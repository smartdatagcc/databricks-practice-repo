**Extract Addresses Data From Files**

**1.Limitations Of CSV File Format In SELECT Statement**

```sql
%python
dfAddressess=spark.read.csv("/Volumes/gizmobox_gr/landing/operational_data/addresses")
display(dfAddressess)
```

**2.Use read_file Function To Overcome Limitations**

```sql
%python
dfAddresses=spark.read.format("csv").option("header", "true").option("delimiter","\t").load("/Volumes/gizmobox_gr/landing/operational_data/addresses")
display(dfAddresses)
```

**3.Create View in Bronze Schema**

```sql
%python
dfAddresses.writeTo("gizmobox_gr.bronze.py_addresses").createOrReplace();

%sql
SELECT * FROM gizmobox_gr.bronze.py_addresses
```
