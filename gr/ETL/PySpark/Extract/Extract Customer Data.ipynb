**1. Query Single Json File**

```sql
%python
dfCustomers=spark.read.format("json").load("/Volumes/gizmobox_gr/landing/operational_data/customers/customers_2024_10.json")
display(dfCustomers);
```

**2.Query Multiple Json Files**

```sql
%python
dfCustomers=spark.read.format("json").load("/Volumes/gizmobox_gr/landing/operational_data/customers/customers_2024_*.json")
display(dfCustomers);
```

**3.Query All The Json Files From Folder**

```sql
%python
dfCustomers=spark.read.json("/Volumes/gizmobox_gr/landing/operational_data/customers")
display(dfCustomers);
```

**4.Select File Meta Data**

```sql
%python
dfMetadata=dfCustomers.select("_metadata.file_path","*")
display(dfMetadata)
```

**5.Registering Files In Unity Catalog Using Views**

```sql
%python
dfMetadata.write.mode("overwrite").saveAsTable("gizmobox_gr.bronze.py_customers");

# Create table with V2
dfMetadata.writeTo("gizmobox_gr.bronze.py_customers").createOrReplace();
```
