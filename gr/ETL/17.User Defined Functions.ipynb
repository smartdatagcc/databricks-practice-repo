**User Defined Functions**

```sql
SELECT 
paymentId,
orderId,
paymetOn,
gizmobox_gr.default.get_payment_status(paymentStatus) paymentStatus,
paymentMethod
 FROM gizmobox_gr.bronze.payments;
```

```sql
CREATE OR REPLACE FUNCTION gizmobox_gr.default.get_payment_status(status int)
RETURNS STRING
RETURN CASE status
WHEN 1 THEN 'Success'
WHEN 2 THEN 'Pending'
WHEN 3 THEN 'Cancelled'
WHEN 4 THEN 'Failed'
ELSE 'Unknown'
END
```
