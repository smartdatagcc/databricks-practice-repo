**CREATE EXTERNAL LOCATION IF NOT EXISTS**

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_practice_ext_dl_circuitboxgr
URL 'abfss://circuitbox-gr@ourdbpracticeextdl.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-practice-ext-sc`)
COMMENT 'External location for databricks_practice_ext_dl_circuitboxgr';
```

**CREATE CATALOG**

```python
%sql
CREATE CATALOG IF NOT EXISTS circuitbox_gr
      MANAGED LOCATION 'abfss://circuitbox-gr@ourdbpracticeextdl.dfs.core.windows.net/'
      COMMENT 'Data Lake house circuitbox catalog'
```

**CREATE SCHEMAS**

```python
%sql
use catalog circuitbox_gr;

CREATE SCHEMA IF NOT EXISTS landing
MANAGED LOCATION 'abfss://circuitbox-gr@ourdbpracticeextdl.dfs.core.windows.net/landing'
COMMENT 'landing schema for circuitbox' ;

CREATE SCHEMA IF NOT EXISTS lakehouse
MANAGED LOCATION 'abfss://circuitbox-gr@ourdbpracticeextdl.dfs.core.windows.net/lakehouse'
COMMENT 'lakehouse schema for circuitbox' ;
```

**CREATE VOLUME**

```python
%sql
USE CATALOG circuitbox_gr;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://circuitbox-gr@ourdbpracticeextdl.dfs.core.windows.net/landing/operational_data'
    COMMENT 'volume for operational data'
```
