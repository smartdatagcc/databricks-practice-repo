**Process Customer Data**

```sql

CREATE OR REFRESH STREAMING TABLE bronze_customers
COMMENT 'Raw customers data'
TBLPROPERTIES('quality' = 'bronze')
AS
SELECT *,'_metadata.file_path' as input_file_path,CURRENT_TIMESTAMP as ingestion_timestamp FROM cloud_files(
    '/Volumes/circuitbox_gr/landing/operational_data/customers',
    'json',
    map('cloudFiles.inferColumnTypes', 'true')
);
```
